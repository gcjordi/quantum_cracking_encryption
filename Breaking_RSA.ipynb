# Toy RSA → Shor → Recovered Private Key

```python
# Portable bootstrap: works from a repo checkout; in Colab it can install from GitHub.
import sys, pathlib, importlib.util
root = pathlib.Path.cwd()
if (root / 'quantum_crypto_lab').exists():
    sys.path.insert(0, str(root))
elif (root.parent / 'quantum_crypto_lab').exists():
    sys.path.insert(0, str(root.parent))
elif importlib.util.find_spec('quantum_crypto_lab') is None:
    %pip install -q 'git+https://github.com/gcjordi/quantum_cracking_encryption.git'
```

```python
from quantum_crypto_lab import ShorSimulator, build_toy_keypair, encrypt_int, decrypt_int
from quantum_crypto_lab.rsa_demo import recover_private_key_from_factors
key = build_toy_keypair(p=5, q=7, e=5)
message = 12
ciphertext = encrypt_int(message, key.public_key)
print('Public key:', key.public_key)
print('Ciphertext:', ciphertext)
```

```python
factored = ShorSimulator().factor(key.n, shots=4096, seed=7)
print('Recovered factors:', factored.factors)
p, q = factored.factors
recovered_private = recover_private_key_from_factors(key.public_key, p, q)
print('Recovered private key:', recovered_private)
print('Recovered message:', decrypt_int(ciphertext, recovered_private))
```

This is a toy demonstration only. Real RSA uses vastly larger moduli and secure padding/constructions; this notebook is not production cryptography.
